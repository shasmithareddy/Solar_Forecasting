```python
import pandas as pd

def analyze_dataset(df, name):
    analysis = {
        "Dataset": name,
        "Number of Rows": df.shape[0],
        "Number of Columns": df.shape[1],
        "Missing Values": "None" if df.isnull().sum().sum() == 0 else "Yes",
        "Basic Features (Temp, Humidity, etc.)": all(col in df.columns for col in ['Air Temperature', 'Humidity']),
        "Lag Features": any('Lag' in col for col in df.columns),
        "Rolling Statistics": any('Rolling' in col for col in df.columns),
        "Interaction Features": any('Interaction' in col for col in df.columns),
        "Fourier Transform Feature": any('Fourier' in col for col in df.columns),
        "Ready for ML Models": "Yes" if any('Lag' in col for col in df.columns) else "No",
        "Ready for DL Models": "Yes" if any('Lag' in col for col in df.columns) else "No",
        "Preprocessing Needed": "Minimal" if any('Lag' in col for col in df.columns) else "High"
    }
    return analysis

# Run analysis on your two datasets
summary = []
summary.append(analyze_dataset(df1, "Approach 1 (Feature Engineered)"))
summary.append(analyze_dataset(df2, "Riyadh NASA (Raw)"))

# Create and display the comparison DataFrame
comparison_df = pd.DataFrame(summary)
display(comparison_df)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select only numeric columns
numeric_df1 = df1.select_dtypes(include='number')

# Plot the correlation matrix for the first 10 numeric columns (to keep it readable)
plt.figure(figsize=(12, 8))
sns.heatmap(numeric_df1.iloc[:, :10].corr(), annot=True, cmap='coolwarm')
plt.title("Dataset 1 - Correlation Heatmap (First 10 Numeric Features)")
plt.show()
```

```python
df1['Datetime'] = pd.to_datetime(df1['Datetime'])
df2['Datetime'] = pd.to_datetime(df2['Datetime'])

print("Dataset 1 Date Range:", df1['Datetime'].min(), "to", df1['Datetime'].max())
print("Dataset 2 Date Range:", df2['Datetime'].min(), "to", df2['Datetime'].max())
```

```python
from statsmodels.tsa.stattools import adfuller

def check_stationarity(series, name):
    result = adfuller(series)
    print(f"ADF Test for {name}:")
    print(f"  p-value = {result[1]:.5f}")
    print("  Stationary" if result[1] < 0.05 else "  Non-stationary")

check_stationarity(df1['Solar Radiation (All Sky)'], "Dataset 1")
check_stationarity(df2['Solar Radiation (All Sky)'], "Dataset 2")
```

```python
df1['Datetime'] = pd.to_datetime(df1['Datetime'])
df2['Datetime'] = pd.to_datetime(df2['Datetime'])

df1.set_index('Datetime')['Solar Radiation (All Sky)'].resample('M').mean().plot(label='Approach 1')
df2.set_index('Datetime')['Solar Radiation (All Sky)'].resample('M').mean().plot(label='NASA Raw')
plt.title("Monthly Avg Solar Radiation Comparison")
plt.legend(); plt.show()
```

```python
cv1 = df1['Solar Radiation (All Sky)'].std() / df1['Solar Radiation (All Sky)'].mean()
cv2 = df2['Solar Radiation (All Sky)'].std() / df2['Solar Radiation (All Sky)'].mean()
print(f"CV - Approach 1: {cv1:.2f}, NASA Raw: {cv2:.2f}")
```

```python
from scipy.stats import zscore
import numpy as np
outliers_df1 = (np.abs(zscore(df1.select_dtypes(include='number'))) > 3).sum()
outliers_df2 = (np.abs(zscore(df2.select_dtypes(include='number'))) > 3).sum()
print("Outlier Count - Approach 1:\n", outliers_df1.sort_values(ascending=False).head())
print("\nOutlier Count - NASA Raw:\n", outliers_df2.sort_values(ascending=False).head())
```
